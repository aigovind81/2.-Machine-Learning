```python
	# Before go into the model 3 things need to do Domain selection 
	# 1. Machine Learning - I/P and O/P  (Simple Linear Regression)
	# 2. Based on user  I/P and O/P it is- Supervised learning 
    # 3. (Classification/Regression ) Based on Numerical value O/P it is Regression 
	# Follow as per the MAPPING STEP BY STEP same steps for all the model.
	
	# Import pandas library (used to handle datasets like Excel (or) CSV-Comma Separated Value) there are 1000 of formula in pandas already loaded we just use it for our job.
	
	# as used for pands as pd short form
import pandas as pd  
```

```python
	# Open the Anaconda Navigator file and open Jupyter notebook and open the new folder for this M/L and use it and give proper naming as well as.
	# Read CSV file and store it in 'dataset'
dataset = pd.read_csv(r'D:\HOPE AI\2. AI Course\Week 3. MACHINE LEARNING REGRESSION\1. Simple Linear Regression\Salary_Data.csv')
```

```python
	# Display the dataset (table view) check the file weather it is loaded correct or not always make sure before go to next set check and go 
dataset
```

```python
	# Select independent variable (it is a input variable or feature) 
	# Here: In this case we use Years Experience → HR want to use to predict salary from the csv file input 
    # check it weather it is properly load or not
#need to start from the first if not found indefication error
independent = dataset[["Years Experience"]]
independent
```

```python
	# Select dependent variable - salary is the O/P  (target/output) 
	# Here: Salary → what we want to predict for different years of experience 
dependent = dataset[["Salary"]]
```

```python
	# Import function to split dataset into 70% Training Set & 30% Testing Set 
    # By using the method sklearn is the MODEL - selection is the FUNCTION always use underscore for the line to consider as one line 
from sklearn.model_selection import train_test_split
```

```python
	# Split data: divide the csv file into Train Set and Test Set
	# 70% for training, 30% for testing
	# random_state=0 ensures same split every time 
    # Below is the formula for need to use for all the model need to know 
X_train, X_test, Y_train, Y_test = train_test_split(independent, dependent, test_size=0.30, random_state=0)
    # Until here all the things same for ALL Algorithms
```

```python
	# Hands on model creation learning PHASE -1 
	# Based on 1 I/p and 1 O/P we choose as SLR model 
	# Import Linear Regression model
	# This is the PROCEDURE we need to teach them correctly then only it will give good result and get high marks
from sklearn.linear_model import LinearRegression  
```

```python
	# Create model object for our work
    #  always write as it is
	# Learning phase 
regressor = LinearRegression()
```

```python
	# Train the model using training data (Model EXAM)
regressor.fit(X_train, Y_train)
```

```python
	# Our formula y=mX + C 
	# y = O/P (which forms a straight line by adding all data points)
	# mX = Slope - Weight (Which says about constant distance b/w 2 datapoints)
	# C = Bais - intercept - straight of the straight line (minimum value) 
	# Get slope (weight) → how much salary increases per year 
weight = regressor.coef_
weight
```

```python
	# Get intercept (bias) → base salary when experience = 0
bais = regressor.intercept_
bais
```

```python
	# War B/W actual and predict value 
	# Predict salary for test data
y_pred = regressor.predict(X_test)
```

```python
	# Import R² score function (model accuracy check)
from sklearn.metrics import r2_score  
```

```python
	# Calculate accuracy (0 is worst or near to 1 is GOOD model)
	# by using the r2 to know how well the model is fitted to our work
r_score = r2_score(Y_test, y_pred)
```

```python
	# Display accuracy 
r_score
```

```python
    # value is 0.99951881 near to 1 so it is good model
	# So Save the trained model by using pickle function for further use 
import pickle
```

```python
    # File name to save model as it do not change the name 
filename = "finalized_model_linear.sav"
```

```python
	# Save model into file 
pickle.dump(regressor, open(filename, 'wb'))
```

```python
	# Deployment = Phase - 2 
	# If you want to shows to customer below details is enough 
	# Load the saved model in pickle
loaded_model = pickle.load(open("finalized_model_linear.sav", 'rb'))
```
